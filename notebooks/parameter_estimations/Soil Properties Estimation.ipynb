```python
import pandas as pd
import numpy as np
from rosetta import rosetta
```

## Jablje

```python
clay = 16.77
silt = 13.43 + 17.47 + 24.63
sand = 23.22 + 3.98 + 0.51
print(sand, silt, clay)

bulk_density = 1.55
```

```python
sand + silt + clay
```

```python
data = [
    [sand, silt, clay, bulk_density]
]
```

```python
#model_code = 2 # use sand, silt, clay
model_code = 3 # use sand, silt, clay, bulk_density
mean, stdev, codes = rosetta(3, data)
```

```python
result = pd.DataFrame(np.vstack([mean, stdev]), columns=["theta_r", "theta_s", "alpha", "n", "Ksat", "K0", "Ipar"], index=["mean", "stdev"])
```

```python
print(result)
```
